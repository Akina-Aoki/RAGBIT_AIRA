# Read PDF

```python
from pypdf import PdfReader

reader = PdfReader("data/Netherland_Dwarf_rabbit.pdf")

all_text = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        all_text += text + "\n"
print(all_text)
```
